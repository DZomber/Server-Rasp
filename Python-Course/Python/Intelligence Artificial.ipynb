```python
#perceptron simple
import matplotlib.pyplot as plt


x1 = [0.3, -0.6,-0.1,0.1]
x2 = [0.7, 0.3, -0.8, -0.45]
t = [1, 0, 0, 1]
w1 = 0.8
w2 = -0.5
b = 0.5

yt = []
delta_w1t =[]
delta_w2t =[]
w1t = []
w2t = []
ut = []
errort = []
c =0
while c <= 1:
    a = 1
    for i in range(len(t)):
        print(f'Iteracion {a}')
        u =w1*x1[i] + w2*x2[i]
        print(f'u = {round(u,3)}')
        y = 0 if u < 0 else 1
        yt.append(y)
        print(f'y = {y}')
        error = t[i] - y
        errort.append(error)
        print(f'error = {round(error,2)}')
        if error == 0 and t[i] ==y:
            print(f'u = {round(u,2)}')
            ut.append(round(u,2))
        elif error == 1 and t[i] ==1 and y ==0:
            print(f'recalculando pesos..')
            delta_w1 =b*x1[i]
            delta_w2 =b*x2[i]
            w1 = w1+delta_w1
            w2 = w2+delta_w2
            u =w1*x1[i] + w2*x2[i]
            print(f'u = {round(u,3)}')
            delta_w1t.append(round(delta_w1,2))
            delta_w2t.append(round(delta_w2,2))
            w1t.append(round(w1,2))
            w2t.append(round(w2,2))
            ut.append(round(u,3))
        elif error == -1 and t[i] ==0 and y ==1:
            print(f'recalculando pesos..')
            delta_w1 =-b*x1[i]
            delta_w2 =-b*x2[i]
            w1 = w1+delta_w1
            w2 = w2+delta_w2
            u =w1*x1[i] + w2*x2[i]
            print(f'u = {round(u,3)}')
            delta_w1t.append(round(delta_w1,2))
            delta_w2t.append(round(delta_w2,2))
            w1t.append(round(w1,2))
            w2t.append(round(w2,2))
            ut.append(round(u,3))

        a =a+1
        print(f'w1 = {round(w1,3)}')
        print(f'w2 = {round(w2,3)}')

    c = c+1
    plt.subplot(1,2,1)
    plt.plot(t,'-o')
    plt.plot(yt,'-^')

plt.subplot(1,2,2)
plt.plot(errort,'-^')
print(f'x1 = {x1}')
print(f'x2 = {x2}')
print(f't = {t}')
print(f'y = {yt}')
print(f'error = {errort}')
print(f'delta_w1t = {delta_w1t}')
print(f'delta_w2t = {delta_w2t}')
print(f'w1t = {w1t}')
print(f'w2t = {w2t}')
print(f'ut = {ut}')
```

```python
#perceptron con bias
x1 = [0,0,1,1]
x2 = [0,1,0,1]
t = [0, 0, 0, 1]
w1 = 0.8
w2 = -0.5
b = 0.5
bias = 0
yt = []
delta_w1t =[]
delta_w2t =[]
w1t = []
w2t = []
ut = []
errort = []
biast =[]
c =0
while c <= 6:
    a = 1
    for i in range(len(t)):
        print(f'Iteracion {a}')
        u =w1*x1[i] + w2*x2[i]-bias
        print(f'u = {round(u,3)}')
        y = 0 if u < 0 else 1
        yt.append(y)
        print(f'y = {y}')
        error = t[i] - y
        errort.append(error)
        print(f'error = {round(error,2)}')
        if error == 0 and t[i] ==y:
            print(f'u = {round(u,2)}')
            ut.append(round(u,2))
        elif error == 1 and t[i] ==1 and y ==0:
            print(f'recalculando pesos..')
            delta_w1 =b*x1[i]
            delta_w2 =b*x2[i]
            w1 = w1+delta_w1
            w2 = w2+delta_w2
            bias = bias -b*error
            u =w1*x1[i] + w2*x2[i]-bias
            print(f'u = {round(u,3)}')
            delta_w1t.append(round(delta_w1,2))
            delta_w2t.append(round(delta_w2,2))
            w1t.append(round(w1,2))
            w2t.append(round(w2,2))
            ut.append(round(u,3))
            biast.append(bias)
        elif error == -1 and t[i] ==0 and y ==1:
            print(f'recalculando pesos..')
            delta_w1 =-b*x1[i]
            delta_w2 =-b*x2[i]
            w1 = w1+delta_w1
            w2 = w2+delta_w2
            bias = bias -b*error
            u =w1*x1[i] + w2*x2[i]-bias

            print(f'u = {round(u,3)}')
            delta_w1t.append(round(delta_w1,2))
            delta_w2t.append(round(delta_w2,2))
            w1t.append(round(w1,2))
            w2t.append(round(w2,2))
            ut.append(round(u,3))
            biast.append(bias)

        a =a+1
        print(f'w1 = {round(w1,3)}')
        print(f'w2 = {round(w2,3)}')

    c = c+1
    plt.subplot(1,2,1)
    plt.plot(t,'-o')
    plt.plot(yt,'-^')

plt.subplot(1,2,2)
plt.plot(errort,'-^')
print(f'x1 = {x1}')
print(f'x2 = {x2}')
print(f't = {t}')
print(f'y = {yt}')
print(f'error = {errort}')
print(f'delta_w1t = {delta_w1t}')
print(f'delta_w2t = {delta_w2t}')
print(f'w1t = {w1t}')
print(f'w2t = {w2t}')
print(f'ut = {ut}')
```

```python
import matplotlib.pyplot as plt
import numpy as np  # Se necesita para usar mean()

x = [1, 2, 3, 4]
t = [0.75, 1.53, 2.34, 3.2]
w = 0.5
b = 0.1
c = 0
error_total =[]
while c <= 3:
    suma = 0
    yt = []  # Reiniciar yt en cada iteración
    errort = []  # Reiniciar errort en cada iteración

    for i in range(len(t)):
        print(f'Iteracion {i+1}')
        y = w * x[i]
        yt.append(y)
        print(f'y{i+1} = {round(y,3)}')

        error = t[i] - y
        errort.append(error)  # Mantener la lista de errores
        print(f'error{i+1} = {round(error,2)}')


    mean_error = np.mean(np.abs(errort))
    error_total.append(mean_error)# Calcular el promedio de errores
    print(f'Error medio: {round(mean_error, 4)}')

    for j in range(len(t)):
        suma += errort[j] * x[j]

    delta_w = b * (1 / len(x) * suma)
    print(f'delta_w{c+1} = {round(delta_w,4)}')
    w = w + delta_w
    print(f'w{c+1} = {round(w,4)}')
    print('*' * 20)

    c += 1


# ==== GRAFICAR RESULTADOS ====
plt.subplot(1, 2, 1)
plt.plot(t, '-o', label='Valores esperados')
plt.plot(yt, '-^', label='Valores obtenidos')
plt.xlabel('Patrón de entrada')
plt.ylabel('Salida')
plt.title('Evolución de la salida')
plt.legend()
plt.grid()

plt.subplot(1, 2, 2)
plt.plot(error_total, '-^', label='Error por entrada')  # Ahora errort es una lista válida
plt.xlabel('Patrón de entrada')
plt.ylabel('Error')
plt.title('Evolución del error')
plt.legend()
plt.grid()

plt.show()
```

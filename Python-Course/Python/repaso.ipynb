```python
import matplotlib.pyplot as plt
#Perceptron simple
x1 = [0.3,-0.6,-0.1,0.1]
x2 = [0.7,0.3,-0.8,-0.45]
t = [1,0,0,1]
w1 = 0.8
w2 = -0.5
beta = 0.5
w1_t =[]
w2_t =[]
yt =[]
ut =[]
errort =[]
```

```python
for i in range(len(x1)):
    u = w1*x1[i]+w2*x2[i]

    y = 1 if u >= 0 else 0
    yt.append(y)
    error = t[i]-y
    errort.append(error)
    w1 = round(w1,3)
    w2 = round(w2,3)

    if error ==1 and t[i] ==1 and y ==0 :
        print(f'recalcular pesos w1 {w1} w2 {w2}')
        delta_w1 = beta*x1[i]
        delta_w2 = beta*x2[i]
        w1 =w1+delta_w1
        w2 =w2+delta_w2
        u =w1*x1[i]+w2*x2[i]
        ut.append(round(u,3))
        w1_t.append(round(w1,3))
        w2_t.append(round(w2,3))

    if error ==-1 and t[i] ==0 and y ==1 :
        print(f'recalcular w1 {w1} w2 {w2}')
        delta_w1 = -beta*x1[i]
        delta_w2 = -beta*x2[i]
        w1 =w1+delta_w1
        w2 =w2+delta_w2
        u =w1*x1[i]+w2*x2[i]
        ut.append(round(u,3))
        w1_t.append(w1)
        w2_t.append(w2)

    elif error ==0 and t[i] ==y:
        ut.append(round(u,3))

print(f'U ={ut}')
print(f'Y ={yt}')
print(f'E ={errort}')
print(f'W1 = {w1_t}')
print(f'W2 = {w2_t}')
plt.plot(t,'-o')
plt.plot(yt,'-*')
```

```python
#Network with Bias

x1 = [0,0,1,1]
x2 = [0,1,0,1]
t = [0,0,0,1]

w1 = 0.8
w2 = -0.5
beta = 0.5
bias = 0
'''
w1 = 0.356
w2 = 0.259
beta = 0.05
bias = 0.122
'''
w1_t =[]
w2_t =[]
yt =[]
ut =[]
errort =[]
biast = []
```

```python
contador = 0
while contador <= 6:
    for i in range(len(x1)):
        u = w1*x1[i]+w2*x2[i]-bias

        y = 1 if u >= 0 else 0
        yt.append(y)
        error = t[i]-y
        errort.append(error)
        w1 = round(w1,3)
        w2 = round(w2,3)

        if error ==1 and t[i] ==1 and y ==0 :
            print(f'recalcular pesos w1 {w1} w2 {w2}')
            delta_w1 = beta*x1[i]
            delta_w2 = beta*x2[i]
            w1 =w1+delta_w1
            w2 =w2+delta_w2
            bias = bias -beta*error
            u =w1*x1[i]+w2*x2[i]-bias
            ut.append(round(u,3))
            w1_t.append(round(w1,3))
            w2_t.append(round(w2,3))
            biast.append(bias)

        if error ==-1 and t[i] ==0 and y ==1 :
            print(f'recalcular w1 {w1} w2 {w2}')
            delta_w1 = -beta*x1[i]
            delta_w2 = -beta*x2[i]
            w1 =w1+delta_w1
            w2 =w2+delta_w2
            bias = bias -beta*error
            u =w1*x1[i]+w2*x2[i]-bias
            ut.append(round(u,3))
            w1_t.append(w1)
            w2_t.append(w2)
            biast.append(bias)

        elif error ==0 and t[i] ==y:
            ut.append(round(u,3))
    contador+=1
print(f'U ={ut}')
print(f'Y ={yt}')
print(f'E ={errort}')
print(f'W1 = {w1_t}')
print(f'W2 = {w2_t}')
print(f'Bias = {biast}')
plt.plot(t,'-o')
plt.plot(yt,'-*')
```

```python
#online
x1 = [0.3,-0.6,-0.1,0.1]
x2 = [0.7,0.3,-0.8,-0.45]
w1 = 0.8
w2 = -0.5
beta = 0.5
t = [1,0,0,1]
Errort =[]
yt =[]
# Use range to iterate over the desired number of times (length of t)
for i in range(len(t)):
    u = round((w1*x1[i])+(w2*x2[i]),5)
    y = u
    yt.append(y)
    # Compare elements of lists t and y instead of lists themselves.
    if t[i] ==1 and y <= 0:
        Error = t[i] - y
        Errort.append(Error)
        delta_w1 = beta * Error * x1[i]
        delta_w2 = beta * Error * x2[i]
        w1 = w1 + delta_w1
        w2 = w2 + delta_w2
X2 =round(-w1/w2,2)
print("predicciones: ",yt)
print("Errores: ",Errort)
print("ajustes de pesos: ",delta_w1, delta_w2)
print("pesos finales:",round(w1,4),round(w2,4))
print("x2",X2)
```

```python
#online
x = [1.0,2.0,3.0,4.0]
t = [0.75,1.53,2.34,3.2]
w = 0.5
beta = 0.1
cont = 0
while cont < 6:
    for i in range(len(x)):
        print('Iteracion',i+1)
        y = round(w*x[i],3)
        print(f'y = {y}')
        error = round(t[i]-y,3)
        print(f'error = {error}')
        w =  round(w + beta*x[i]*error,3)
        print(f'w{i+1} = {w}')
    cont += 1
```

```python
x = [1.0,2.0,3.0,4.0]
t = [0.75,1.53,2.34,3.2]
w = 0.5
beta = 0.1
yt =[]
cont =0
while cont < 10:
    n = len(x)
    suma = 0
    errort = []
    for i in range(len(x)):
        print('Iteracion',i+1)
        y = round(w*x[i],3)
        yt.append(y)
        print(f'y = {y}')
        error = round(t[i]-y,3)
        errort.append(error)
        print(f'error = {error}')
    print(n)
    for j in range(len(x)):
        suma += errort[j]*x[j]

    delta_w = round(beta*((1/n)*suma),4)

    w = w + delta_w
    cont +=1
    print(f'delta_w{cont} = {delta_w}')
    print(f'w{cont} ={w}')

    print('*'*20)
```

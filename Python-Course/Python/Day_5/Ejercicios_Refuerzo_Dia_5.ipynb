```python

def devolver_distintos(n1,n2,n3):
    num_max =0
    num_min =0
    num_inter =0
    tres_numeros = [n1,n2,n3]
    num_max = max(tres_numeros)
    num_min = min(tres_numeros)
    suma = n1+n2+n3
    num_inter= 0
    
    if n1 != num_max and n1 != num_min:
        num_inter= n1
    elif n2 != num_max and n2 != num_min:
        num_inter= n2
    elif n3 != num_max and n3 != num_min:
        num_inter= n3
    
    print(f'la suma es {suma}')
    
    if suma > 15:
        return f'El numero maximo es {num_max}'
    if suma <10:
        return f'El numero minimo es {num_min}'
    if suma in range(11,16):
        return f'El numero intermedio es {num_inter}'
    
devolver_distintos(6,4,5)
```

```python
def cantidad_ceros(*args):
    
    for i in range(len(args) - 1):
        if args[i] == 0 and args[i + 1] == 0:
            return True
    return False

# Testing the function
print(cantidad_ceros(6, 1, 5, 1, 0, 3, 0, 1))  # Output: False
print(cantidad_ceros(6, 1, 5, 1, 0, 0, 3, 1))  # Output: True
```

```python
def numeros_primos(*args):
    cantidad_primos = 0
    for arg in args:
        if arg == 0 or arg == 1:
            pass
        elif arg % 2 == 0:
            print(f'el numero {arg} es par')
        else: 
            print(f'el numero {arg} es impar') 
            cantidad_primos += 1
    print(f'la cantidad de numeros primos es {cantidad_primos}')
numeros_primos(6, 1, 5, 1, 0, 3, 0, 1,3,9,7,5)
```

```python
def letras_unicas(palabra):
    mi_set = set()
    for letra in palabra:
        mi_set.add(letra)
    mi_list = list(mi_set)
    mi_list.sort()
    return mi_list
letras_unicas('entretenido')
    
    
```

```python
def ceros_vecinos(*args):
    contador = 0
    for arg in args:
        if contador +1 == len(args):
            return False
        if args[contador] == 0 and args[contador + 1] == 0:
            return True
        else: contador += 1
    return False
ceros_vecinos(6, 1, 5, 1, 0, 3, 0, 1,3,9,7,5)
```

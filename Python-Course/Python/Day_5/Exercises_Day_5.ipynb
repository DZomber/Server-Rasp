```python
def suma_cuadrados(*args):
    suma = 0
    for arg in args:
        suma += arg**2
    return suma
print(suma_cuadrados(1,2,3,4,5))
```

```python
def suma_absolutos(*args):
    suma = 0
    for arg in args:
        suma += abs(arg)
    return suma
print(suma_absolutos(1,-2,3,-4,5))
```

```python
def numero_persona(nombre,*args):
    suma_numeros = 0
    for arg in args:
        suma_numeros += arg
    return f'{nombre}, la suma de tus numeros es {suma_numeros}'

print(numero_persona('kerim',1,2,3,4,5))
        
```

```python
def cantidad_atributos(**kwargs):
    cantidad = 0
    for parametros in kwargs.items():
        cantidad += 1
    return cantidad
print(cantidad_atributos(a=1,b=2,c=3))
```

```python
def lista_atributos(**kwargs):
    lista =[]
    for valor in kwargs.values():
        lista.append(valor)
    return lista
print(lista_atributos(a=1,b=2,c=3))
```

```python
class Personaje:

    def __init__(self,cantidad_flechas):
        self.cantidad_flechas = cantidad_flechas
        print(f'tienes {cantidad_flechas} flechas')
    def lanzar_flechas(self):
        self.cantidad_flechas = self.cantidad_flechas -1
        print('Lanzaste una flecha\nte quedan {} flechas '.format(self.cantidad_flechas))
        
x = Personaje(10)

x.lanzar_flechas()
```

```python
class Vertebrado:
    vertebrado = True

class Ave(Vertebrado):
    tiene_pico = True
    def poner_huevos(self):
        print("Poniendo huevos")

class Reptil(Vertebrado):
    venenoso = True

class Pez(Vertebrado):
    def nadar(self):
        print("Nadando")
    def poner_huevos(self):
        print("Poniendo huevos")

class Mamifero(Vertebrado):
    def caminar(self):
        print("Caminando")
    def amamantar(self):
        print("Amamantando crías")

class Ornitorrinco(Pez,Reptil,Ave,Mamifero):
    pass
perry = Ornitorrinco()
perry.poner_huevos()
perry.tiene_pico = True
perry.venenoso = True
perry.nadar()
perry.caminar()
perry.amamantar()


```

```python
class Mago():
    def atacar(self):
        print("Ataque mágico")

class Arquero():
    def atacar(self):
        print("Lanzamiento de flecha")

class Samurai():
    def atacar(self):
        print("Ataque con katana")
personajes = [Mago(), Arquero(), Samurai()]

for personaje in personajes:
    personaje.atacar()
```

```python
class Mago():
    def defender(self):
        print("Escudo mágico")

class Arquero():
    def defender(self):
        print("Esconderse")

class Samurai():
    def defender(self):
        print("Bloqueo")
def personaje_defender(defender):
    defender.defender()
personaje_defender(Samurai())
personaje_defender(Arquero())
personaje_defender(Mago())
```

```python
class Libro():
    def __init__(self, titulo, autor, cantidad_paginas):
        self.titulo = titulo
        self.autor = autor
        self.cantidad_paginas = cantidad_paginas
        
    def __str__(self):
        return f'"{self.titulo}", de {self.autor}'
libro = Libro("The hungry games", "Susane Collines", 300)
print(libro)
```

```python
class Libro():
    def __init__(self, titulo, autor, cantidad_paginas):
        self.titulo = titulo
        self.autor = autor
        self.cantidad_paginas = cantidad_paginas

    def __len__(self):
        return self.cantidad_paginas
libro = Libro("The hungry games", "Susane Collines", 300)
print(len(libro))
```

```python
class Libro():
    def __init__(self, titulo, autor, cantidad_paginas):
        self.titulo = titulo
        self.autor = autor
        self.cantidad_paginas = cantidad_paginas

    def __del__(self):
        print( "Libro eliminado")
libro = Libro("The hungry games", "Susane Collines", 300)
del libro
```

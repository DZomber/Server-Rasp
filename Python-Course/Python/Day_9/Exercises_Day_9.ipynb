```python
import re

def verificar_email(email):
    email =input("Digite su email: ")
    patron = r'@\w+\.com'
    chequear = re.search(patron, email)
    if chequear:
        print("Ok")
    else:
        print("La dirección de email es incorrecta")
verificar_email("<EMAIL>")
```

```python
import re

def verificar_saludo(frase):
    
    patron = 'Hola' 
    busqueda = re.search(patron, frase)
    if busqueda:
        print("Ok")
    else:
        print("No has saludado")
        
verificar_saludo('Hola como has estado')
```

```python
import re

def verificar_cp(cp):
    patron = r'\w{2}\d{4}'
    chequear = re.search(patron, cp)
    if chequear:
        print("Ok")
    else:
        print("El código postal ingresado no es correcto")

verificar_cp('XX1234')

```

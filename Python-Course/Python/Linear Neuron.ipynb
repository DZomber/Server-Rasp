```python


x = [ 1.0, 2.0, 3.0, 4.0]
t = [ 0.75, 1.53, 2.34,3.2]
w = 0.5
beta = 0.1
yt =[]
Et = []
wt =[]
for i in range(0, len(x)):
    y = w*x[i]
    yt.append(y)
    E = t[i]-y 
    Et.append(E)
    
    w += (beta*x[i]*E)
    wt.append(w)
print(f'Y = {yt}')
print(f't = {t}')
print(f'E = {Et}')
print(f'Wt ={wt}')
print('*'*25)

    
    
    
```

```python
x = [1.0, 2.0, 3.0, 4.0]
t = [0.75, 1.53, 2.34, 3.2]
w = 0.5
beta = 0.1

def y_total(w, x):
    return [w * x_i for x_i in x]

def et_total(yt, t):
    return [t_i - y_i for t_i, y_i in zip(t, yt)]

def delta(w, Et):

    return beta * (1/4) * sum(E * x_i for E, x_i in zip(Et, x))

# Iteraciones del aprendizaje
con = 1
while con < 5:
    print(f'Iteración = {con}')
    
    yt = y_total(w, x)
    Et = et_total(yt, t)
    delta_w1 = delta(w,Et)
    print(f'y = {yt}')
    print(f'E = {Et}')
    print(f'delta = {delta_w1:.5f}')
    w += delta_w1
    print(f'w = {w:.5f}')


    print('*' * 25, '\n')
    
    con += 1



    
```
